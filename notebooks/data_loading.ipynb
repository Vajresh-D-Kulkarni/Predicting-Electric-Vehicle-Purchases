```python
def find_value_counts(df: pd.DataFrame) -> dict:

    value_counts = {}
    for column in df.columns:
        if df[column].dtype == "str":
            value_counts[column] = df[column].value_counts()
    return value_counts
```

```python
category_value_counts = find_value_counts(train_data)
```

```python
print(list((column, values) for column, values in category_value_counts.items()))
```
